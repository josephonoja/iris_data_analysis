# DATA VISUALIZATION


## 1: Distribuion of flower species

```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns

sns.set_theme(style="whitegrid")

plt.figure(
    figsize=(8,5)
)

sns.countplot(
    data=df,
    x="species",
    hue="species",
    palette=("red", "green", "blue"),
    legend=False,
    edgecolor="black"
)

plt.title("Distribution of Iris Flower Species")
plt.xlabel("Species")
plt.ylabel("Number of Flowers")

plt.tight_layout()
plt.savefig("../images/species_distribution.png", dpi=300)
plt.show()
```

### Observation

The chart shows that each Iris species contains **50 flowers**. this confirms that the dataset is balanced, with an equal number of observations for each species.

### Interpretation

The balanced distribution indicates that no species is overrepresented.This allows fair comparison among the three Iris species during the analysis.

### Business Insight

A balanced dataset is valuable for researchers and data analysts because it reduces bias and supports reliable comparisons between species.

### Recommendation

Researchers can confidently use this dataset to compare characteristics of the three Iris species is equally represented.


## 2: Average Sepal Length by Species

Preparing the data by creating a variable

```python
average_sepal_length = (
    df.groupby("species")["sepal_length"]
      .mean()
      .reset_index()
)
average_sepal_length
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8,5))

sns.barplot(
    data=average_sepal_length,
    x="species",
    y="sepal_length",
    palette=("red", "green", "blue"),
    hue="species",
    legend=False,
    edgecolor="black"
)

plt.title("Average Sepal Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Sepal Length (cm)")

plt.tight_layout()

plt.savefig("../images/average_sepal_length_by_species.png", dpi=300)

plt.show()
```

### Observation

**Iris_virginica** has the highest average sepal length among the three species.

**Iris_setosa** has the lowest average sepal length.

The average sepal length increases from **Iris_setosa** to **Iris_versicolor** and is highest in **Iris_virginica**.

### Interpretation

The analysis shows that the three Iris species differ in their average sepal lengths. the Iris_virginica generally has longer sepals compared to Iris_versicolor and Iris_setosa, while Iris_setosa has the shortest average sepal length. This suggests that sepal length maybe a useful feature for distinguishing and comparing the characteristics of different Iris species

### Business Insight

This analysis helps botanists, researchers and scientists compare physical differences between Iris species and identify useful features for species classification and further studies.

### Recommendation

Researchers can further investigate biological or environmental factors that may influence differences in sepal length among Iris species

### 6: Distribution of Sepal Length

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="sepal_length",
    bins=10,
    color="red",
    edgecolor="black"
)

plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig("../images/sepal_length_distribution.png", dpi=300)

plt.show()
```

### Observation

The distribution of sepal length is concentrated within the middle ranges of values.

Most flowers sepal lengths between approximately **5cm and 7cm**.

Very few flowers have extremely short or extremely long sepal length.

### Interpretation

The histogram shows that sepal length are not evenly distributed across the dataset. Most Iris flowers have moderate sepal lengths, while very small and very large sepal length occur less frequently.

### Business Insight

Understanding the distribution of sepal length helps botanists and researchers identify common growth patterns and detect unsual observations that may require further investigations.

### Recommendation

Researchers should investigate the factors that contribute to variations in sepal length and examine whether this differences are associated with specific Iris species or environmental conditions.

### 7: Relationship Between Sepal Length and Petal Length

```python
import matplotlib.pyplot as plt
import seaborn as sns 

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="sepal_length",
    y="petal_length",
    hue="species",
    palette=("red", "green", "blue"),
    s=80,
    edgecolor="black"
)

plt.title("Relationship Between Sepal Length and Petal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")

plt.tight_layout()

plt.savefig("../images/sepal_length_vs_petal_length.png", dpi=300)

plt.show()
```

### Observation

The scatter plot shows a positive relationship between sepal length and petal length.

As sepal length increases, petal length also tends to increase.

The Iris species form distinct clusters based on their sepal and petal measurements.

### Interpretation

The analysis indicates a positive relationship between sepal length and petal length across the Iris species. This suggests that flowers with longer sepals genrally tend to have longer petals, making these measurements useful for distinguishing between species.

### Business Insight

This analysis helps botanists and researchers understand the relationship between flower characteristics, supporting species identification and botanical research.

### Recommendation

Researchers should further investigate the biological, genetic and environmental factors that may influence the relationship between sepal length and petal length among Iris species.
